```python
import pandas as pd
```

```python
df = pd.read_excel("C:\\Users\\Legion\\Downloads\\SFT_Contextual_10000_updated.xlsx")
print("Dataset loaded successfully!")
df
```

```python
# Calculate duplicate SFT samples

duplicate_sft_count = (
    df.duplicated(
        subset=[
            "instruction",
            "input",
            "output"
        ]
    )
    .sum()
)

duplicate_rate = (
    duplicate_sft_count
    / len(df)
    * 100
)

print(
    "Duplicate SFT samples:",
    duplicate_sft_count
)

print(
    "Duplicate rate:",
    round(
        duplicate_rate,
        2
    ),
    "%"
)
```

```python
# Set the overall quality score

quality_score = 50

print(
    "Quality score:",
    quality_score
)
```

```python
# Assess training readiness

def assess_training_readiness(
    quality_score,
    duplicate_rate
):

    if duplicate_rate > 50:

        return (
            "Not Recommended for Training"
        )

    elif duplicate_rate > 20:

        return (
            "Needs Duplicate Reduction"
        )

    elif quality_score >= 90:

        return (
            "Ready for Training"
        )

    elif quality_score >= 75:

        return (
            "Ready After Minor Review"
        )

    else:

        return (
            "Needs Quality Improvement"
        )
```

```python
# Get the training readiness result

training_readiness = (
    assess_training_readiness(
        quality_score,
        duplicate_rate
    )
)

print(
    "Training readiness:",
    training_readiness
)
```

```python
# Determine training risk level

def get_risk_level(
    duplicate_rate
):

    if duplicate_rate > 50:

        return "Critical"

    elif duplicate_rate > 20:

        return "High"

    elif duplicate_rate > 5:

        return "Medium"

    else:

        return "Low"
```

```python
# Get the training risk level

risk_level = get_risk_level(
    duplicate_rate
)

print(
    "Training risk level:",
    risk_level
)
```

```python
# Generate training recommendations

recommendations = []

if duplicate_rate > 50:

    recommendations.append(
        "Do not use the dataset directly for model training."
    )

    recommendations.append(
        "Review duplicate SFT samples before training."
    )

    recommendations.append(
        "Remove exact duplicate samples."
    )

    recommendations.append(
        "Increase instruction and output diversity."
    )

    recommendations.append(
        "Create or collect more unique SFT samples."
    )

elif duplicate_rate > 20:

    recommendations.append(
        "Reduce duplicate samples before training."
    )

    recommendations.append(
        "Review dataset diversity."
    )

else:

    recommendations.append(
        "The dataset can proceed to training review."
    )

    recommendations.append(
        "Perform a final manual quality review."
    )

for number, recommendation in enumerate(
    recommendations,
    start=1
):

    print(
        number,
        recommendation
    )
```

```python
# Create the final SFT dataset assessment report

final_report = pd.DataFrame({
    "Metric": [
        "Total Samples",
        "Overall Quality Score",
        "Quality Status",
        "Duplicate SFT Samples",
        "Duplicate Rate",
        "Training Risk Level",
        "Training Readiness"
    ],
    "Result": [
        len(df),
        f"{quality_score}/100",
        "Poor",
        duplicate_sft_count,
        f"{duplicate_rate:.2f}%",
        risk_level,
        training_readiness
    ]
})

final_report
```

```python
# Create the recommendation report

recommendation_report = pd.DataFrame({
    "Recommendation Number": range(
        1,
        len(recommendations) + 1
    ),
    "Recommendation": recommendations
})

recommendation_report
```

```python
# Display the complete final assessment

print("SFT DATASET FINAL ASSESSMENT REPORT")

print()

print(
    "Total Samples:",
    len(df)
)

print(
    "Overall Quality Score:",
    f"{quality_score}/100"
)

print(
    "Quality Status:",
    "Poor"
)

print(
    "Duplicate SFT Samples:",
    duplicate_sft_count
)

print(
    "Duplicate Rate:",
    f"{duplicate_rate:.2f}%"
)

print(
    "Training Risk Level:",
    risk_level
)

print(
    "Training Readiness:",
    training_readiness
)

print()

print("RECOMMENDATIONS")

print()

for number, recommendation in enumerate(
    recommendations,
    start=1
):

    print(
        f"{number}. {recommendation}"
    )
```

```python
# Save the final assessment report

report_path = (
    "../output/reports/"
    "final_sft_assessment_report.csv"
)

final_report.to_csv(
    report_path,
    index=False
)

print(
    "Final assessment report saved successfully"
)
```

```python
# Save the final recommendations

recommendation_path = (
    "../output/reports/"
    "final_training_recommendations.csv"
)

recommendation_report.to_csv(
    recommendation_path,
    index=False
)

print(
    "Final recommendations saved successfully"
)
```
