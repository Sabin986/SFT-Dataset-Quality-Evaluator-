```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("C:\\Users\\Legion\\Downloads\\SFT_Contextual_10000_updated.xlsx")
print("Dataset loaded successfully!")
df
```

```python
# Load the quality scoring result

score_path = "../output/reports/quality_scoring_result.csv"

scoring_result = pd.read_csv(score_path)

scoring_result
```

```python
# Load the quality check summary

summary_path = "../output/reports/quality_penalty_details.csv"

quality_summary = pd.read_csv(summary_path)

quality_summary
```

```python
# Create a bar chart for quality issue counts

plt.figure(figsize=(12, 6))

plt.bar(
    quality_summary["Quality Check"],
    quality_summary["Issue Count"]
)

plt.title(
    "SFT Dataset Quality Issues"
)

plt.xlabel(
    "Quality Check"
)

plt.ylabel(
    "Issue Count"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()
```

```python
# Save the quality issues chart

chart_path = (
    "../output/charts/"
    "quality_issues_chart.png"
)

plt.figure(figsize=(12, 6))

plt.bar(
    quality_summary["Quality Check"],
    quality_summary["Issue Count"]
)

plt.title(
    "SFT Dataset Quality Issues"
)

plt.xlabel(
    "Quality Check"
)

plt.ylabel(
    "Issue Count"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.savefig(
    chart_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Quality issues chart saved successfully"
)
```

```python
# Calculate domain counts

domain_counts = (
    df["domain"]
    .value_counts()
)

domain_counts
```

```python
# Create a domain distribution chart

plt.figure(figsize=(10, 6))

plt.bar(
    domain_counts.index,
    domain_counts.values
)

plt.title(
    "SFT Dataset Domain Distribution"
)

plt.xlabel(
    "Domain"
)

plt.ylabel(
    "Number of Samples"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()
```
