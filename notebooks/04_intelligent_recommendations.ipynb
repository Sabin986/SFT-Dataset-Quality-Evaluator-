```python
import pandas as pd
```

```python
# Load the quality check summary

summary_path = "../output/reports/quality_check_summary.csv"

quality_summary = pd.read_csv(summary_path)

quality_summary
```

```python
# Load the quality scoring result

score_path = "../output/reports/quality_scoring_result.csv"

scoring_result = pd.read_csv(score_path)

scoring_result
```

```python
# Create a dictionary of quality issue counts

issue_counts = dict(
    zip(
        quality_summary["Quality Check"],
        quality_summary["Issue Count"]
    )
)

issue_counts
```

```python
# Generate recommendations based on quality issues

def generate_recommendations(
    issue_counts,
    total_samples
):

    recommendations = []

    if issue_counts.get(
        "Missing Values",
        0
    ) > 0:

        recommendations.append(
            "Review and handle missing values before fine-tuning."
        )

    if issue_counts.get(
        "Empty Text Values",
        0
    ) > 0:

        recommendations.append(
            "Remove or complete samples with empty text fields."
        )

    if issue_counts.get(
        "Duplicate IDs",
        0
    ) > 0:

        recommendations.append(
            "Assign unique IDs to all dataset samples."
        )

    if issue_counts.get(
        "Duplicate SFT Samples",
        0
    ) > 0:

        recommendations.append(
            "Review duplicate SFT samples and improve dataset diversity."
        )

    if issue_counts.get(
        "Very Short Instructions",
        0
    ) > 0:

        recommendations.append(
            "Review very short instructions because they may be unclear."
        )

    if issue_counts.get(
        "Very Short Inputs",
        0
    ) > 0:

        recommendations.append(
            "Review very short inputs because they may lack context."
        )

    if issue_counts.get(
        "Very Short Outputs",
        0
    ) > 0:

        recommendations.append(
            "Review very short outputs because they may be incomplete."
        )

    if issue_counts.get(
        "Very Long Outputs",
        0
    ) > 0:

        recommendations.append(
            "Review very long outputs for unnecessary or repetitive content."
        )

    if issue_counts.get(
        "Empty Domains",
        0
    ) > 0:

        recommendations.append(
            "Assign a valid domain to samples with missing domains."
        )

    if issue_counts.get(
        "Empty Sources",
        0
    ) > 0:

        recommendations.append(
            "Add source information to samples with missing sources."
        )

    if len(recommendations) == 0:

        recommendations.append(
            "No major quality issues were detected."
        )

        recommendations.append(
            "The dataset is suitable for further review and fine-tuning."
        )

    return recommendations
```

```python
# Set the total number of samples

total_samples = 10000

print(
    "Total samples:",
    total_samples
)
```

```python
# Generate dataset recommendations

recommendations = generate_recommendations(
    issue_counts,
    total_samples
)

recommendations
```

```python
# Display recommendations

print(
    "SFT DATASET RECOMMENDATIONS"
)

print()

for number, recommendation in enumerate(
    recommendations,
    start=1
):

    print(
        number,
        recommendation
    )
```

```python
# Create a recommendation table

recommendation_table = pd.DataFrame({
    "Recommendation Number": range(
        1,
        len(recommendations) + 1
    ),
    "Recommendation": recommendations
})

recommendation_table
```

```python
# Save recommendations

output_path = (
    "../output/reports/"
    "intelligent_recommendations.csv"
)

recommendation_table.to_csv(
    output_path,
    index=False
)

print(
    "Recommendations saved successfully"
)
```
