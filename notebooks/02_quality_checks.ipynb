```python
import pandas as pd
import numpy as np
```

```python
# Load the SFT dataset


df = pd.read_excel("C:\\Users\\Legion\\Downloads\\SFT_Contextual_10000_updated.xlsx")
print("Dataset loaded successfully")
df
```

```python
# Display the dataset shape

print("Total rows:", df.shape[0])

print("Total columns:", df.shape[1])
```

```python
# Display the dataset columns

print(df.columns.tolist())
```

```python
# Check whether all required columns are available

required_columns = [
    "id",
    "domain",
    "instruction",
    "input",
    "output",
    "source"
]

missing_columns = []

for column in required_columns:
    if column not in df.columns:
        missing_columns.append(column)

if len(missing_columns) == 0:
    print("All required columns are available")
else:
    print("Missing columns:")
    print(missing_columns)
```

```python
# Count missing values in each column

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Check empty values in text columns

text_columns = [
    "domain",
    "instruction",
    "input",
    "output",
    "source"
]

empty_counts = {}

for column in text_columns:

    empty_count = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    empty_counts[column] = empty_count

print(empty_counts)
```

```python
# Count exact duplicate rows

duplicate_rows = df.duplicated().sum()

print("Total duplicate rows:", duplicate_rows)
```

```python
# Check duplicate IDs

duplicate_ids = df["id"].duplicated().sum()

print("Duplicate IDs:", duplicate_ids)
```

```python
# Check duplicate instructions

duplicate_instructions = (
    df["instruction"]
    .duplicated()
    .sum()
)

print("Duplicate instructions:", duplicate_instructions)
```

```python
# Check duplicate outputs

duplicate_outputs = (
    df["output"]
    .duplicated()
    .sum()
)

print("Duplicate outputs:", duplicate_outputs)
```

```python
# Check duplicate SFT samples

duplicate_samples = (
    df.duplicated(
        subset=[
            "instruction",
            "input",
            "output"
        ]
    )
    .sum()
)

print("Duplicate SFT samples:", duplicate_samples)
```

```python
# Count unique values in important columns

print("Unique instructions:", df["instruction"].nunique())

print("Unique inputs:", df["input"].nunique())

print("Unique outputs:", df["output"].nunique())
```

```python
# Display unique instruction values

df["instruction"].unique()
```

```python
# Count occurrences of each instruction

instruction_counts = (
    df["instruction"]
    .value_counts()
)

instruction_counts
```

```python
# Calculate instruction length using word count

df["instruction_length"] = (
    df["instruction"]
    .astype(str)
    .str.split()
    .str.len()
)

df[
    [
        "instruction",
        "instruction_length"
    ]
].head()
```

```python
# Calculate input length using word count

df["input_length"] = (
    df["input"]
    .astype(str)
    .str.split()
    .str.len()
)

df[
    [
        "input",
        "input_length"
    ]
].head()
```

```python
# Calculate output length using word count

df["output_length"] = (
    df["output"]
    .astype(str)
    .str.split()
    .str.len()
)

df[
    [
        "output",
        "output_length"
    ]
].head()
```

```python
# Display text length statistics

length_statistics = df[
    [
        "instruction_length",
        "input_length",
        "output_length"
    ]
].describe()

length_statistics
```

```python
# Count very short instructions

minimum_instruction_length = 3

short_instructions = (
    df["instruction_length"]
    < minimum_instruction_length
).sum()

print(
    "Very short instructions:",
    short_instructions
)
```

```python
# Count very short inputs

minimum_input_length = 3

short_inputs = (
    df["input_length"]
    < minimum_input_length
).sum()

print(
    "Very short inputs:",
    short_inputs
)
```

```python
# Count very short outputs

minimum_output_length = 5

short_outputs = (
    df["output_length"]
    < minimum_output_length
).sum()

print(
    "Very short outputs:",
    short_outputs
)
```

```python
# Count very long outputs

maximum_output_length = 500

long_outputs = (
    df["output_length"]
    > maximum_output_length
).sum()

print(
    "Very long outputs:",
    long_outputs
)
```

```python
# Display very short output samples

df[
    df["output_length"] < minimum_output_length
][
    [
        "id",
        "instruction",
        "input",
        "output",
        "output_length"
    ]
].head(10)
```

```python
# Display all unique domain values

unique_domains = (
    df["domain"]
    .astype(str)
    .str.strip()
    .sort_values()
    .unique()
)

print("Total unique domains:", len(unique_domains))

print(unique_domains)
```

```python
# Count samples in each domain

domain_counts = (
    df["domain"]
    .astype(str)
    .str.strip()
    .value_counts()
)

domain_counts
```

```python
# Check empty domain values

empty_domains = (
    df["domain"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("Empty domains:", empty_domains)
```

```python
# Check domain values with leading or trailing spaces

domain_values = (
    df["domain"]
    .astype(str)
)

domains_with_extra_spaces = (
    domain_values
    .ne(domain_values.str.strip())
    .sum()
)

print(
    "Domains with extra spaces:",
    domains_with_extra_spaces
)
```

```python
# Calculate domain distribution percentage

domain_percentage = (
    df["domain"]
    .astype(str)
    .str.strip()
    .value_counts(
        normalize=True
    )
    * 100
)

domain_percentage.round(2)
```

```python
# Identify the largest and smallest domains

largest_domain = (
    domain_counts.idxmax()
)

largest_domain_count = (
    domain_counts.max()
)

smallest_domain = (
    domain_counts.idxmin()
)

smallest_domain_count = (
    domain_counts.min()
)

print(
    "Largest domain:",
    largest_domain
)

print(
    "Largest domain sample count:",
    largest_domain_count
)

print(
    "Smallest domain:",
    smallest_domain
)

print(
    "Smallest domain sample count:",
    smallest_domain_count
)
```

```python
# Calculate the domain balance ratio

domain_balance_ratio = (
    smallest_domain_count
    / largest_domain_count
)

print(
    "Domain balance ratio:",
    round(
        domain_balance_ratio,
        2
    )
)
```

```python
# Display all unique source values

unique_sources = (
    df["source"]
    .astype(str)
    .str.strip()
    .unique()
)

print("Total unique sources:", len(unique_sources))

print(unique_sources)
```

```python
# Count samples from each source

source_counts = (
    df["source"]
    .astype(str)
    .str.strip()
    .value_counts()
)

source_counts
```

```python
# Check empty source values

empty_sources = (
    df["source"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

print("Empty sources:", empty_sources)
```

```python
# Check source values with extra spaces

source_values = (
    df["source"]
    .astype(str)
)

sources_with_extra_spaces = (
    source_values
    .ne(source_values.str.strip())
    .sum()
)

print(
    "Sources with extra spaces:",
    sources_with_extra_spaces
)
```

```python
# Check the number of samples from each source

print(
    "Largest source:",
    source_counts.idxmax()
)

print(
    "Largest source count:",
    source_counts.max()
)

print(
    "Smallest source:",
    source_counts.idxmin()
)

print(
    "Smallest source count:",
    source_counts.min()
)
```

```python
# Calculate total missing values

total_missing_values = df.isnull().sum().sum()

print("Total missing values:", total_missing_values)
```

```python
# Calculate total empty text values

total_empty_values = (
    empty_counts["domain"]
    + empty_counts["instruction"]
    + empty_counts["input"]
    + empty_counts["output"]
    + empty_counts["source"]
)

print("Total empty text values:", total_empty_values)
```

```python
# Create the quality check summary

quality_summary = pd.DataFrame({
    "Quality Check": [
        "Missing Values",
        "Empty Text Values",
        "Duplicate IDs",
        "Duplicate SFT Samples",
        "Very Short Instructions",
        "Very Short Inputs",
        "Very Short Outputs",
        "Very Long Outputs",
        "Empty Domains",
        "Domains With Extra Spaces",
        "Empty Sources",
        "Sources With Extra Spaces"
    ],
    "Issue Count": [
        total_missing_values,
        total_empty_values,
        duplicate_ids,
        duplicate_samples,
        short_instructions,
        short_inputs,
        short_outputs,
        long_outputs,
        empty_domains,
        domains_with_extra_spaces,
        empty_sources,
        sources_with_extra_spaces
    ]
})

quality_summary
```

```python
# Calculate issue percentage

quality_summary["Issue Percentage"] = (
    quality_summary["Issue Count"]
    / len(df)
    * 100
)

quality_summary["Issue Percentage"] = (
    quality_summary["Issue Percentage"]
    .round(2)
)

quality_summary
```

```python
# Assign quality status to each check

def get_issue_status(issue_count):

    if issue_count == 0:
        return "Passed"

    elif issue_count <= len(df) * 0.01:
        return "Minor Issue"

    elif issue_count <= len(df) * 0.05:
        return "Moderate Issue"

    else:
        return "Major Issue"


quality_summary["Status"] = (
    quality_summary["Issue Count"]
    .apply(get_issue_status)
)

quality_summary
```

```python
# Display the final quality check report

print("SFT DATASET QUALITY CHECK REPORT")

print()

print("Total samples:", len(df))

print()

print(
    quality_summary.to_string(
        index=False
    )
)
```

```python
# Save the quality check summary

output_path = (
    "../output/reports/"
    "quality_check_summary.csv"
)

quality_summary.to_csv(
    output_path,
    index=False
)

print(
    "Quality check summary saved successfully"
)
```
